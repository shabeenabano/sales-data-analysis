# PRODUCT ANALYSIS

# 20. Revenue by Product Line

```python
# Calculate the total revenue for each product line
product_sales = df.groupby('Product line')['Total'].sum()

# Display revenue by product line 
print(product_sales)
```

# Visualization

```python
# Create a bar chart for revenue by product line
plt.figure(figsize=(10, 6))
product_sales.plot(kind='bar', color=['skyblue', 'orange', 'green', 'red', 'purple', 'gold'])

# Add title and labels
plt.title('Total Revenue by Product Line')
plt.xlabel('product Line')
plt.ylabel('Total Revenue')

# Rotate labels for better readability
plt.xticks(rotation=45)

# Display the chart
plt.show()
```

# 21. Quantity Sold by Product Line

# 35. Final Business Insights

This section summarizes the major findings from the sales analysis.

```python
# Find the product line with the highest total revenue
highest_revenue_product = product_sales.idxmax()
highest_revenue_value = product_sales.max()

# Find the product line with the highest quantity sold
highest_quantity_product = product_quantity.idxmax()
highest_quantity_value = product_quantity.max()

# Find the product line with the highest gross income
highest_revenue_product = product_income.idxmax()
highest_income_value = product_income.max()

# Find the branch with the highest revenue
highest_revenue_branch = branch_sales.idxmax()
highest_branch_value = branch_sales.max()

# Find the city with the highest revenue
highest_revenue_city = city_sales.idxmax()
highest_city_value = city_sales.max()

# Find the most profitable product line based on gross income
most_profitable_product = product_income.idxmax()

# Find the product line with the highest average rating
highest_rating_product = product_rating.idxmax()
highest_rating_value = product_rating.max()

# Find the most popular payment method based on transaction count 
most_used_payment = df['Payment'].value_counts().idxmax()
most_used_payment_count = df['Payment'].value_counts().max()

# Find the gross income with the highest revenue
highest_income_product = product_income.idxmax()
highest_income_value = product_income.max()

# Display the key business insights
print("===== FINAL BUSINESS INSIGHTS =====\n")


print(f"1. Highest revenue product line: {highest_revenue_product}")
print(f" Revenue: {highest_revenue_value:.2f}\n")

print(f"2. Highest quantity sold product line: {highest_quantity_product}")
print(f" Quantity sold: {highest_quantity_value}\n")

print(f"3. Highest gross income product line: {highest_income_product}")
print(f" Gross income: {highest_income_value:.2f}\n")

print(f"4. Highest revenue branch: {highest_revenue_branch}")
print(f" Revenue: {highest_branch_value:.2f}\n")

print(f"5. Highest revenue city: {highest_revenue_city}")
print(f" Revenue: {highest_city_value:.2f}\n")

print(f"6. Most profitable product line: {most_profitable_product}\n")

print(f"7. Highest rated product line: {highest_rating_product}")
print(f" Average rating: {highest_rating_value:.2f}\n")

print(f"8. Most frequently used payment method: {most_used_payment}")
print(f" Number of transactions: {most_used_payment_count}")
```
